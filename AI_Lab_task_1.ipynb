```python
def calculator(a,b,operator):
  if operator == "+":
    return a + b
  elif operator == "-":
    return a - b
  elif operator == "*":
    return a * b
  elif operator == "/":
    if b != 0:
      return a / b
    else:
      print("Cannot divide by zero")
  else:
    print("Invalid operator")
print(calculator(10,5,"+"))
print(calculator(10,5,"-"))
print(calculator(10,5,"*"))
print(calculator(10,5,"/"))
```

```python
#Student marks management system
# Is program mein hum list + dictionary + function + loop + if-else use karenge.
students = [
    {"name" : "Ali" , "marks" : 90},
    {"name" : "tariq" , "marks" : 80},
    {"name" : "Sara" , "marks" : 45},
    {"name" : "Jawad" , "marks" : 60}
]
def check_marks(marks):
  if marks >= 50:
    return "Pass"
  else:
    return "Fail"
for student in students:
  name = student["name"]
  marks = student["marks"]
  result = check_marks(marks)
  print("Name: " , name)
  print("Marks: " , marks)
  print("Result: " , result)
  print("----------------")

```
